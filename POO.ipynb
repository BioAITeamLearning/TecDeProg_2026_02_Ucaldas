# 1. Clases y objetos

## 1.1 Teoría

- **Clase**: el molde. **Objeto**: una cosa hecha con ese molde. Igual que en Java.
- El bloque se delimita con **indentación**, no con llaves.
- No se declara `public`, `private` ni el tipo de los atributos.
- El primer parámetro de todo método es **`self`**, y hay que escribirlo.

### De Java a Python

| En Java | En Python |
|---|---|
| `public class Estudiante {` | `class Estudiante:` |
| `{ }` delimitan el bloque | la **indentación** delimita el bloque |
| `public Estudiante(...) { }` | `def __init__(self, ...):` |
| `this` | `self` (explícito, siempre) |
| `new Estudiante("Ana")` | `Estudiante("Ana")` (sin `new`) |
| `private String nombre;` | *(no se declara: nace al asignarlo)* |

## 1.2 Definir una clase

````python
class Estudiante:
    """Un estudiante del curso."""

    def __init__(self, nombre, codigo):
        self.nombre = nombre        # el atributo NACE aqui, al asignarlo
        self.codigo = codigo
````

````python
# Crear objetos: sin "new"
ana = Estudiante("Ana", "2026001")
luis = Estudiante("Luis", "2026002")

print(ana.nombre, "-", ana.codigo)
print(luis.nombre, "-", luis.codigo)
````

````python
# Cada objeto tiene SUS propios atributos
ana.nombre = "Ana María"

print(ana.nombre)
print(luis.nombre)   # luis no cambio
````

## 1.3 `self`: el parámetro que Java te escondía

En Java, cuando escribes `ana.saludar()`, la JVM le pasa al método una referencia oculta al objeto:
eso es `this`. Python hace lo mismo, pero **no lo esconde**: te obliga a declararlo como primer
parámetro.

````python
class Estudiante:
    def __init__(self, nombre, codigo):
        self.nombre = nombre
        self.codigo = codigo

    def saludar(self):                       # self SIEMPRE es el primer parametro
        return f"Hola, soy {self.nombre}"
````

````python
ana = Estudiante("Ana", "2026001")

print(ana.saludar())              # la forma normal
print(Estudiante.saludar(ana))    # lo mismo, con el self explicito
````

```{important}
Olvidar `self` es el error nº 1 de quien viene de Java:

    def saludar():                 # ❌ falta self
        return f"Hola, soy {self.nombre}"

    ana.saludar()
    # TypeError: saludar() takes 0 positional arguments but 1 was given

Y la otra mitad de la trampa: dentro del método, `nombre` a secas **no existe**. Siempre
`self.nombre`. En Java podías omitir el `this.` y funcionaba; aquí no.
```

# 2. Atributos

## 2.1 De instancia y de clase

- **De instancia**: viven en cada objeto (`self.nombre`). Cada objeto tiene el suyo.
- **De clase**: viven en la clase y son **compartidos** por todos los objetos. Es el `static` de Java.

````python
class Estudiante:
    universidad = "Universidad de Caldas"   # DE CLASE: compartido (static)
    matriculados = 0                        # DE CLASE: contador comun

    def __init__(self, nombre):
        self.nombre = nombre                # DE INSTANCIA: propio de cada objeto
        Estudiante.matriculados += 1
````

````python
ana = Estudiante("Ana")
luis = Estudiante("Luis")

print(ana.nombre, "|", luis.nombre)          # distintos
print(ana.universidad, "|", luis.universidad) # el mismo
print("matriculados:", Estudiante.matriculados)
````

````python
# Un atributo de clase se cambia en la CLASE, y lo ven todos los objetos
Estudiante.universidad = "U. de Caldas — Facultad de IA e Ingenierías"

print(ana.universidad)
print(luis.universidad)
````

```{warning}
**La trampa.** Si asignas sobre el **objeto** en vez de sobre la clase, no cambias el atributo
compartido: creas uno nuevo, de instancia, que lo tapa solo para ese objeto.
```

````python
ana.universidad = "Otra universidad"   # crea un atributo DE INSTANCIA en ana

print("ana  :", ana.universidad)
print("luis :", luis.universidad)               # sigue con el de la clase
print("clase:", Estudiante.universidad)         # la clase no se toco
````

# 3. Métodos

## 3.1 Métodos de instancia

Son los normales: reciben `self` y trabajan con los datos del objeto.

````python
class Estudiante:
    def __init__(self, nombre, notas):
        self.nombre = nombre
        self.notas = notas

    def promedio(self):
        if not self.notas:
            return 0.0
        return sum(self.notas) / len(self.notas)

    def aprobo(self):
        return self.promedio() >= 3.0        # un metodo puede llamar a otro con self
````

````python
ana = Estudiante("Ana", [4.5, 3.8, 4.0])
print(f"{ana.nombre}: {ana.promedio():.2f} | aprobo: {ana.aprobo()}")
````

## 3.2 `@staticmethod` y `@classmethod`

| En Java | En Python |
|---|---|
| `public static boolean validar(x)` | `@staticmethod` |
| *(constructor alternativo)* | `@classmethod` |

- **`@staticmethod`**: no recibe `self` ni la clase. Es una utilidad que vive dentro de la clase.
- **`@classmethod`**: recibe la **clase** (`cls`) en vez de la instancia. Sirve sobre todo para
  crear objetos de otra manera.

````python
class Estudiante:
    def __init__(self, nombre, notas=None):
        self.nombre = nombre
        self.notas = list(notas) if notas else []

    @staticmethod
    def codigo_valido(codigo):
        """No usa self: es una utilidad que vive en la clase."""
        return isinstance(codigo, str) and len(codigo) == 7 and codigo.isdigit()

    @classmethod
    def desde_texto(cls, linea):
        """Constructor alternativo: recibe la CLASE (cls), no la instancia.
        Espera un texto de la forma 'Ana:4.5,3.8,4.0'"""
        nombre, notas = linea.split(":")
        return cls(nombre, [float(n) for n in notas.split(",")])
````

````python
print(Estudiante.codigo_valido("2026001"))   # True
print(Estudiante.codigo_valido("20A6001"))   # False
````

````python
ana = Estudiante.desde_texto("Ana:4.5,3.8,4.0")
print(ana.nombre, ana.notas)
````

# 4. Constructores

## 4.1 `__init__`

El constructor **no lleva el nombre de la clase**: siempre se llama `__init__`. Y solo puede haber
**uno** — en Python **no existe la sobrecarga de métodos**.

````python
class Estudiante:
    def __init__(self, nombre, codigo, semestre=1):   # semestre tiene valor por defecto
        self.nombre = nombre
        self.codigo = codigo
        self.semestre = semestre
````

````python
a = Estudiante("Ana", "2026001")        # usa el valor por defecto
b = Estudiante("Luis", "2026002", 5)

print(a.nombre, a.semestre)
print(b.nombre, b.semestre)
````

```{note}
**Aquí está el reemplazo de la sobrecarga.** En Java escribirías dos constructores:
`Estudiante(String, String)` y `Estudiante(String, String, int)`. En Python la segunda definición
**borraría** la primera, así que se usa **un solo constructor con valores por defecto**.
```

## 4.2 La trampa del argumento por defecto mutable

La más traicionera de todas. El valor por defecto se evalúa **una sola vez**, al definir la
función — así que todas las instancias terminan compartiendo la *misma* lista.

````python
class CursoRoto:
    def __init__(self, inscritos=[]):     # ❌ NUNCA una lista como valor por defecto
        self.inscritos = inscritos

a = CursoRoto()
b = CursoRoto()
a.inscritos.append("Ana")

print("a:", a.inscritos)
print("b:", b.inscritos, "  <- b se contamino!")
````

````python
class CursoOK:
    def __init__(self, inscritos=None):   # ✅ el patron correcto
        self.inscritos = list(inscritos) if inscritos else []

a = CursoOK()
b = CursoOK()
a.inscritos.append("Ana")

print("a:", a.inscritos)
print("b:", b.inscritos)
````

## 4.3 `@dataclass`: el POJO en una línea

Cuando una clase solo guarda datos, en Java escribes constructor, getters, `equals`, `hashCode` y
`toString` — o usas un `record`. En Python hay un atajo.

````python
from dataclasses import dataclass

@dataclass
class Nota:
    materia: str
    valor: float
    creditos: int = 3        # valor por defecto
````

````python
n1 = Nota("Tecnicas de Programacion", 4.5)
n2 = Nota("Tecnicas de Programacion", 4.5)

print(n1)                  # __repr__ generado automaticamente
print("n1 == n2 ->", n1 == n2)   # __eq__ generado automaticamente
````

```{note}
Las anotaciones `materia: str` **no son tipos obligatorios**: Python no las verifica al ejecutar.
Son documentación (y ayuda para el editor). En `@dataclass` cumplen además la función de decirle
al decorador cuáles son los campos.
```

# 5. Encapsulamiento

## 5.1 Python no tiene `private`

Esta es la diferencia que más incomoda al que viene de Java, así que vamos de frente:

> **En Python no existe `private`, `protected` ni `public`. Todo es accesible siempre.**

Lo que existe es una **convención**: un atributo que empieza con guion bajo (`_promedio`) le dice
al resto del mundo *"esto es interno, no lo toques"*. Nada te impide tocarlo. Es un acuerdo entre
adultos, no una barrera del compilador.

````python
class Estudiante:
    def __init__(self, nombre, promedio):
        self.nombre = nombre
        self._promedio = promedio      # "privado" por convencion
````

````python
ana = Estudiante("Ana", 4.2)

print(ana._promedio)     # Python te deja leerlo

ana._promedio = 99.9     # y tambien te deja romperlo
print(ana._promedio)
````

## 5.2 `@property`: el getter que no parece getter

Es el equivalente del *getter*, pero **se usa como si fuera un atributo**: sin paréntesis.

````python
class Estudiante:
    def __init__(self, nombre, promedio):
        self.nombre = nombre
        self._promedio = promedio

    @property
    def promedio(self):
        """Getter: se usa como atributo, sin parentesis."""
        return self._promedio

    @promedio.setter
    def promedio(self, valor):
        """Setter: valida antes de asignar."""
        if not 0.0 <= valor <= 5.0:
            raise ValueError("El promedio debe estar entre 0.0 y 5.0")
        self._promedio = valor
````

````python
ana = Estudiante("Ana", 4.2)

print(ana.promedio)     # sin parentesis: parece un atributo

ana.promedio = 4.8      # pero pasa por el setter
print(ana.promedio)
````

````python
try:
    ana.promedio = 7.5      # el setter lo bloquea
except ValueError as e:
    print("Bloqueado:", e)
````

```{tip}
**La regla práctica de Python es la contraria a la de Java.**

- En Java: atributo `private` + getter/setter *desde el principio*, por si acaso.
- En Python: atributo **público** desde el principio, y **si algún día** necesitas validar, lo
  conviertes en `@property` — sin romper a nadie que ya estuviera escribiendo `ana.promedio`.

Por eso verán muchísimo código Python sin un solo getter. No es descuido: es que no hacían falta.
```

# 6. Métodos especiales (*dunder*)

Son los métodos con doble guion bajo (*double underscore*). Le enseñan a tu clase a comportarse
como los tipos del lenguaje: `len(x)`, `x == y`, `for i in x`, `print(x)`.

| Lo que quieres escribir | El método que debes implementar | En Java sería |
|---|---|---|
| `print(obj)` | `__str__` | `toString()` |
| `obj` en el notebook | `__repr__` | `toString()` |
| `len(obj)` | `__len__` | `size()` |
| `if obj:` | `__bool__` | `!isEmpty()` |
| `a == b` | `__eq__` | `equals()` |
| `x in obj` | `__contains__` | `contains()` |
| `obj[i]` | `__getitem__` | `get(i)` |
| `for x in obj:` | `__iter__` | `implements Iterable` |

## 6.1 `__repr__` y `__str__`

````python
class Estudiante:
    def __init__(self, nombre, codigo):
        self.nombre = nombre
        self.codigo = codigo

ana = Estudiante("Ana", "2026001")
print(ana)      # <__main__.Estudiante object at 0x...>  <- inutil
````

````python
class Estudiante:
    def __init__(self, nombre, codigo):
        self.nombre = nombre
        self.codigo = codigo

    def __repr__(self):
        """Para el programador: se ve en el notebook y dentro de listas."""
        return f"Estudiante({self.nombre!r}, {self.codigo!r})"

    def __str__(self):
        """Para el usuario final: lo usa print()."""
        return f"{self.nombre} ({self.codigo})"
````

````python
ana = Estudiante("Ana", "2026001")

print(ana)          # usa __str__
print(repr(ana))    # usa __repr__
print([ana])        # dentro de una lista SIEMPRE se usa __repr__
````

```{tip}
Si solo vas a escribir uno, escribe **`__repr__`**: cuando `__str__` no existe, Python usa
`__repr__` como respaldo. Al revés no funciona.
```

## 6.2 Los protocolos: aquí está su `interface`

En Java, para que una clase se pueda recorrer con un `for-each` tiene que declarar
`implements Iterable<T>`. En Python **no se declara nada**: basta con implementar los métodos que
el protocolo pide.

````python
class Curso:
    def __init__(self, nombre):
        self.nombre = nombre
        self._inscritos = []

    def inscribir(self, estudiante):
        self._inscritos.append(estudiante)

    def __repr__(self):
        return f"Curso({self.nombre!r}, {len(self._inscritos)} inscritos)"

    def __len__(self):                  # len(c)      <- size()
        return len(self._inscritos)

    def __bool__(self):                 # if c:       <- !isEmpty()
        return len(self._inscritos) > 0

    def __contains__(self, nombre):     # "Ana" in c  <- contains()
        return nombre in self._inscritos

    def __getitem__(self, i):           # c[0]        <- get(0)
        return self._inscritos[i]

    def __iter__(self):                 # for x in c: <- Iterable
        return iter(self._inscritos)
````

````python
tec = Curso("Tecnicas de Programacion")
for nombre in ["Ana", "Luis", "Sara"]:
    tec.inscribir(nombre)

print(tec)
print("len(tec)     ->", len(tec))
print("bool(tec)    ->", bool(tec))
print("'Ana' in tec ->", "Ana" in tec)
print("tec[0]       ->", tec[0])
````

````python
for estudiante in tec:
    print("  -", estudiante)

print("curso vacio es falso ->", bool(Curso("Vacio")))
````

```{note}
**Fíjate en lo que acaba de pasar.** No escribimos `tec.size()` ni `tec.isEmpty()` ni
`tec.contains("Ana")`. Escribimos `len(tec)`, `if tec:` y `"Ana" in tec` — la **misma sintaxis**
que usarías con una lista o un diccionario de Python. Tu clase se comporta como las del lenguaje
porque implementa los mismos métodos, no porque lo haya declarado.
```

## 6.3 `__eq__`: el `equals()` de Java

````python
class Estudiante:
    def __init__(self, nombre, codigo):
        self.nombre = nombre
        self.codigo = codigo

    def __repr__(self):
        return f"Estudiante({self.nombre!r}, {self.codigo!r})"

    def __eq__(self, otro):
        if not isinstance(otro, Estudiante):
            return NotImplemented
        return self.codigo == otro.codigo      # dos estudiantes son el mismo si comparten codigo
````

````python
a = Estudiante("Ana", "2026001")
b = Estudiante("Ana María", "2026001")
c = Estudiante("Luis", "2026002")

print("a == b ->", a == b)    # True: mismo codigo
print("a == c ->", a == c)    # False
print("a is b ->", a is b)    # False: son objetos distintos
````

```{warning}
**En Java es al revés.** Allá `==` compara referencias y `.equals()` compara contenido.
En Python `==` compara **contenido** (usa tu `__eq__`) e `is` compara **identidad**.
```

# 7. Herencia

## 7.1 `extends` se escribe con paréntesis

| En Java | En Python |
|---|---|
| `class Estudiante extends Persona` | `class Estudiante(Persona):` |
| `super(nombre, cedula);` | `super().__init__(nombre, cedula)` |
| `@Override` | *(no existe; no hace falta)* |
| `abstract class` | `class X(ABC)` con `@abstractmethod` |

Vamos a refactorizar: `Estudiante` y `Docente` comparten datos, así que sacamos una clase base.

````python
class Persona:
    def __init__(self, nombre, cedula):
        self.nombre = nombre
        self.cedula = cedula

    def presentarse(self):
        return f"Soy {self.nombre}"

    def __repr__(self):
        return f"{type(self).__name__}({self.nombre!r})"
````

````python
class Estudiante(Persona):
    def __init__(self, nombre, cedula, codigo):
        super().__init__(nombre, cedula)    # llama al __init__ del padre
        self.codigo = codigo
````

````python
class Docente(Persona):
    def __init__(self, nombre, cedula, departamento):
        super().__init__(nombre, cedula)
        self.departamento = departamento
````

## 7.2 Sobrescribir un método

Se redefine con el mismo nombre. No hace falta `@Override`. Y `super()` sirve para **reutilizar**
lo que ya hacía el padre en vez de repetirlo.

````python
class Estudiante(Persona):
    def __init__(self, nombre, cedula, codigo):
        super().__init__(nombre, cedula)
        self.codigo = codigo

    def presentarse(self):                       # sobrescribe el del padre
        base = super().presentarse()             # reutiliza lo que ya hacia
        return f"{base}, estudiante con código {self.codigo}"
````

````python
class Docente(Persona):
    def __init__(self, nombre, cedula, departamento):
        super().__init__(nombre, cedula)
        self.departamento = departamento

    def presentarse(self):
        return f"{super().presentarse()}, docente de {self.departamento}"
````

````python
ana = Estudiante("Ana", "1053...", "2026001")
jp = Docente("Johan", "1057...", "Sistemas e Informática")

print(ana.presentarse())
print(jp.presentarse())
````

# 8. Polimorfismo

## 8.1 *Duck typing*: polimorfismo sin `interface`

> *"Si camina como pato y hace cuac como pato, entonces es un pato."*

En Java, para que una función acepte tanto un `Estudiante` como un `Docente`, ambos tendrían que
compartir una superclase o `implements` una interfaz. En Python **no hace falta**: la función
llama al método, y si el objeto lo tiene, funciona.

````python
def pasar_lista(personas):
    """No pregunta de que tipo es cada uno: solo les pide que se presenten."""
    for p in personas:
        print(" -", p.presentarse())
````

````python
class Invitado:
    """No hereda de Persona. Ni siquiera se entero de que existe."""
    def __init__(self, nombre, institucion):
        self.nombre = nombre
        self.institucion = institucion

    def presentarse(self):
        return f"Soy {self.nombre}, invitado de {self.institucion}"
````

````python
ana = Estudiante("Ana", "1053...", "2026001")
jp = Docente("Johan", "1057...", "Sistemas e Informática")
externo = Invitado("Sara", "UAM")

pasar_lista([ana, jp, externo])   # los tres funcionan, sin interface comun
````

```{note}
`Invitado` **no hereda de `Persona`**. Funciona igual porque implementa `presentarse()`.
Eso es *duck typing*: en Java **declaras** que cumples un contrato; en Python **lo cumples** y ya.
```

## 8.2 Clases abstractas: cuando sí quieres el contrato explícito

A veces sí conviene obligar. `abc.ABC` es el equivalente de `abstract class` / `interface`, y
Python **falla** si intentas instanciar una clase que no implementó todos los métodos.

````python
from abc import ABC, abstractmethod

class Evaluable(ABC):                  # ~ interface / abstract class
    @abstractmethod
    def nota_final(self):
        """Toda clase evaluable debe saber calcular su nota final."""
````

````python
class Taller(Evaluable):
    def __init__(self, puntos, sobre):
        self.puntos = puntos
        self.sobre = sobre

    def nota_final(self):
        return round(5.0 * self.puntos / self.sobre, 2)
````

````python
class Parcial(Evaluable):
    def __init__(self, respuestas_ok, total):
        self.respuestas_ok = respuestas_ok
        self.total = total

    def nota_final(self):
        return round(5.0 * self.respuestas_ok / self.total, 2)
````

````python
# Polimorfismo: la misma llamada, comportamientos distintos
evaluaciones = [Taller(38, 40), Parcial(17, 20)]

for e in evaluaciones:
    print(f"{type(e).__name__:10s} -> {e.nota_final()}")
````

````python
class Quiz(Evaluable):
    pass          # se le olvido implementar nota_final

try:
    Quiz()
except TypeError as e:
    print("Error al instanciar:", e)
````

```{tip}
**¿Cuándo usar `ABC` y cuándo *duck typing*?** En este curso, casi siempre *duck typing*: es lo
idiomático y lo que verán en el código de las próximas unidades. `ABC` se justifica cuando diseñas
una jerarquía que **otras personas** van a extender y quieres que el error salga temprano y claro.
```

# 9. Interfaces y clases abstractas

## 9.1 Java tiene dos mecanismos; Python, uno

En Java, `interface` y `abstract class` son cosas distintas con reglas distintas. En Python **ambas
son lo mismo**: una clase que hereda de `ABC`. La diferencia deja de estar en el lenguaje y pasa a
estar en **cómo la escribes**.

| | `interface` (Java) | `abstract class` (Java) | En Python |
|---|---|---|---|
| ¿Puede tener estado (atributos)? | No | Sí | Sí, si tú se lo pones |
| ¿Puede traer código hecho? | Solo `default` | Sí | Sí, si tú se lo pones |
| ¿Cuántas se pueden heredar? | Varias | **Una sola** | **Varias** |
| Cómo se declara | `implements A, B` | `extends A` | `class C(A, B):` |

O sea: en Python una **interfaz pura** es simplemente una `ABC` en la que *decidiste* no poner
implementación. Nadie te obliga.

## 9.2 Clase abstracta *con* implementación

Esto es lo que en Java sería un `abstract class`: obliga a las hijas a implementar `nota()`, pero
además les **regala** `aporte()` ya hecho. Es el patrón *método plantilla*: el padre define el
esqueleto y las hijas rellenan el hueco.

````python
from abc import ABC, abstractmethod

class Evaluacion(ABC):
    def __init__(self, peso):
        self.peso = peso                      # una ABC SI puede tener estado

    @abstractmethod
    def nota(self):
        """Cada evaluacion sabe calcular su propia nota. Obligatorio implementarlo."""

    def aporte(self):
        """Metodo CONCRETO: las hijas lo heredan tal cual, no lo reescriben."""
        return round(self.nota() * self.peso / 100, 2)
````

````python
class Taller(Evaluacion):
    def __init__(self, peso, puntos, sobre):
        super().__init__(peso)
        self.puntos, self.sobre = puntos, sobre

    def nota(self):
        return round(5.0 * self.puntos / self.sobre, 2)
````

````python
class Parcial(Evaluacion):
    def __init__(self, peso, correctas, total):
        super().__init__(peso)
        self.correctas, self.total = correctas, total

    def nota(self):
        return round(5.0 * self.correctas / self.total, 2)
````

````python
evaluaciones = [Taller(30, 38, 40), Parcial(70, 17, 20)]

for e in evaluaciones:
    print(f"{type(e).__name__:8s} nota={e.nota()}  peso={e.peso}%  aporte={e.aporte()}")

print("definitiva:", round(sum(e.aporte() for e in evaluaciones), 2))
````

## 9.3 Interfaz pura

La misma herramienta, pero sin estado ni implementación: solo la lista de lo que hay que saber
hacer. Esto es el `interface` de Java.

````python
class Exportable(ABC):
    """Interfaz pura: solo dice QUE hay que saber hacer, no COMO."""

    @abstractmethod
    def a_dict(self):
        """Retorna los datos del objeto como diccionario."""
````

````python
class Imprimible(ABC):
    """Otra interfaz pura, independiente de la anterior."""

    @abstractmethod
    def encabezado(self):
        """Retorna una linea de texto para mostrar en pantalla."""
````

## 9.4 `implements A, B` es herencia múltiple

Aquí está la diferencia que más se siente. En Java una clase **extiende una sola** clase pero
puede **implementar varias** interfaces. En Python no hay dos verbos: todo es herencia, y puedes
heredar de varias.

```java
// Java
class Estudiante extends Persona implements Exportable, Imprimible { ... }
```

````python
class Persona:
    def __init__(self, nombre, cedula):
        self.nombre, self.cedula = nombre, cedula
````

````python
# Python: una sola lista. La primera suele ser la clase base "de verdad",
# y las demas funcionan como interfaces.
class Estudiante(Persona, Exportable, Imprimible):
    def __init__(self, nombre, cedula, codigo):
        super().__init__(nombre, cedula)
        self.codigo = codigo

    def a_dict(self):                      # obligatorio por Exportable
        return {"nombre": self.nombre, "cedula": self.cedula, "codigo": self.codigo}

    def encabezado(self):                  # obligatorio por Imprimible
        return f"{self.codigo} — {self.nombre}"
````

````python
ana = Estudiante("Ana", "1053...", "2026001")

print(ana.encabezado())
print(ana.a_dict())
print("es Exportable?", isinstance(ana, Exportable))
print("es Imprimible?", isinstance(ana, Imprimible))
````

````python
# Si falta implementar alguna, Python no te deja instanciar
class Incompleto(Persona, Exportable, Imprimible):
    def a_dict(self):
        return {}
    # falta encabezado()

try:
    Incompleto("X", "0")
except TypeError as e:
    print("Error:", e)
````

## 9.5 El orden importa: el MRO

Si heredas de varias clases, ¿cuál gana cuando dos definen el mismo método? Python resuelve eso
con el **MRO** (*Method Resolution Order*): una lista, calculada una sola vez, que dice en qué
orden se busca.

Y aquí viene la sorpresa para quien viene de Java: **`super()` no significa "mi clase padre"**.
Significa **"la siguiente clase en el MRO"**, que puede ser una hermana y no una ancestra.

````python
class Base:
    def describir(self):
        return "Base"

class A(Base):
    def describir(self):
        return "A -> " + super().describir()

class B(Base):
    def describir(self):
        return "B -> " + super().describir()

class C(A, B):
    def describir(self):
        return "C -> " + super().describir()
````

````python
print(C().describir())
````

````python
# El super() de A llamo a B, que NO es su padre: es la siguiente en el MRO de C
print([clase.__name__ for clase in C.__mro__])
````

```{note}
`A` hereda de `Base`, no de `B`. Pero al usarla dentro de `C(A, B)`, el `super()` de `A` va a `B`.
Por eso en herencia múltiple **siempre** se usa `super()` y nunca `Base.metodo(self)`: `super()`
respeta el MRO y llamar a la clase por su nombre lo rompe.
```

## 9.6 *Mixins*: interfaces que además traen comportamiento

Un **mixin** es una clase pensada para ser heredada *junto a* otra, que aporta un comportamiento
suelto. No se instancia sola y no representa un "es un". Es la forma idiomática en Python de lo
que en Java harías con una interfaz de métodos `default`.

Por convención se nombran terminando en `Mixin`.

````python
import json

class ExportaJSONMixin:
    """Aporta a_json() a cualquier clase que ya sepa hacer a_dict()."""

    def a_json(self):
        return json.dumps(self.a_dict(), ensure_ascii=False)
````

````python
class Docente(Persona, Exportable, ExportaJSONMixin):
    def __init__(self, nombre, cedula, departamento):
        super().__init__(nombre, cedula)
        self.departamento = departamento

    def a_dict(self):
        return {"nombre": self.nombre, "departamento": self.departamento}
````

````python
jp = Docente("Johan", "1057...", "Sistemas e Informática")

print(jp.a_json())        # metodo que no escribimos en Docente: vino del mixin
````

## 9.7 Propiedades abstractas

También puedes obligar a que exista un **atributo calculado**, no solo un método. Se combinan los
dos decoradores, y `@property` va arriba.

````python
class Actividad(ABC):

    @property
    @abstractmethod
    def descripcion(self):
        """Toda actividad debe poder describirse."""
````

````python
class Laboratorio(Actividad):
    def __init__(self, tema):
        self.tema = tema

    @property
    def descripcion(self):
        return f"Laboratorio de {self.tema}"
````

````python
lab = Laboratorio("estructuras de datos")
print(lab.descripcion)      # sin parentesis: es una propiedad
````

## 9.8 `Protocol`: la interfaz que no hay que declarar

Todo lo anterior es **nominal**: para ser `Exportable` tienes que *heredar* de `Exportable`.
Igual que en Java.

Pero Python tiene una segunda vía, más fiel a su propia filosofía: `typing.Protocol`. Con ella,
un objeto cumple la interfaz **por tener los métodos**, sin heredar de nada. Es *duck typing*,
pero verificable.

````python
from typing import Protocol, runtime_checkable

@runtime_checkable
class Presentable(Protocol):
    """Cualquier cosa que sepa presentarse cumple este protocolo."""

    def presentarse(self) -> str: ...
````

````python
class Invitado:
    """NO hereda de Presentable. Ni sabe que existe."""

    def __init__(self, nombre, institucion):
        self.nombre, self.institucion = nombre, institucion

    def presentarse(self) -> str:
        return f"Soy {self.nombre}, invitado de {self.institucion}"
````

````python
sara = Invitado("Sara", "UAM")

print(sara.presentarse())
print("cumple Presentable?", isinstance(sara, Presentable))   # True, sin heredar nada
````

```{tip}
**¿Cuál uso?**

| Situación | Herramienta |
|---|---|
| Quiero **obligar** a mis subclases y compartirles código hecho | `ABC` con métodos concretos (§9.2) |
| Quiero un contrato puro y que el error salte al instanciar | `ABC` solo con `@abstractmethod` (§9.3) |
| Quiero agregar un comportamiento suelto a varias clases | *mixin* (§9.6) |
| Quiero aceptar cualquier objeto que "sepa hacer X", sin pedirle que herede | `Protocol` (§9.8) |
| Es código mío, corto, y confío en quien lo llama | *duck typing* a secas (§8.1) |
```

# 11. Errores frecuentes del que viene de Java

| # | El error | Qué pasa |
|---|---|---|
| 1 | Olvidar `self` en la firma del método | `TypeError: ... takes 0 positional arguments but 1 was given` |
| 2 | Escribir `nombre` en vez de `self.nombre` dentro del método | Crea una variable local y la descarta; el atributo no cambia |
| 3 | Escribir `this` | No existe: es `self` |
| 4 | Definir dos métodos con el mismo nombre esperando sobrecarga | El segundo **borra** al primero. Se usa un solo método con valores por defecto |
| 5 | `def __init__(self, xs=[])` | Todas las instancias comparten la **misma** lista |
| 6 | Buscar `private` | No existe. Convención `_atributo` y, si hace falta validar, `@property` |
| 7 | Usar `==` esperando comparar referencias | En Python `==` compara contenido; la identidad es `is` |
| 8 | `respaldo = mi_lista` creyendo que copia | Es otro nombre para el mismo objeto. La copia es `.copy()` |
| 9 | Usar `new` | No existe: `Estudiante("Ana")` |
| 10 | Poner `@Override` | No existe; basta redefinir el método |
