## 3 · ¿Qué imprime?

Para cada celda, **escribe la salida antes de ejecutarla**.

````python
# 3.1
datos = [4, 7, 2, 9, 2, 5]
total = 0
for i, n in enumerate(datos):
    if n == 2:
        continue
    if n > 8:
        break
    total += n
print(total, i)
````

````python
# 3.2
inventario = {"lapiz": 3, "cuaderno": 1}
inventario["borrador"] = inventario.get("borrador", 0) + 5
inventario["lapiz"] += 2
del inventario["cuaderno"]

print(len(inventario))
print(sorted(inventario))
print(inventario.get("cuaderno", "agotado"))
````

````python
# 3.3
class Carrito:
    def __init__(self, items=[]):
        self.items = items
    def agregar(self, x):
        self.items.append(x)

a = Carrito()
b = Carrito()
a.agregar("pan")
print(a.items, b.items)
````

````python
# 3.4
class Empleado:
    def __init__(self, nombre):
        self.nombre = nombre
    def saludo(self):
        return f"Soy {self.nombre}"

class Jefe(Empleado):
    def __init__(self, nombre, area):
        super().__init__(nombre)
        self.area = area
    def saludo(self):
        return super().saludo() + f", jefe de {self.area}"

print(Empleado("Ana").saludo())
print(Jefe("Luis", "Sistemas").saludo())
print(isinstance(Jefe("Luis", "Sistemas"), Empleado))
````

````python
# 3.5
class Caja:
    def __init__(self, cosas):
        self.__cosas = list(cosas)
    def __len__(self):
        return len(self.__cosas)
    def __contains__(self, x):
        return x in self.__cosas
    def __iter__(self):
        return iter(tuple(self.__cosas))

c = Caja(["tuerca", "tornillo", "clavo"])
print(len(c))
print("clavo" in c)
print([x.upper() for x in c])
print(bool(Caja([])))
````
