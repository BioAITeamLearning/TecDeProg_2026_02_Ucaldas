# 1. Lo que hace cada una

| | Pila | Cola | Cola circular |
|---|---|---|---|
| **Principio** | LIFO — el último en entrar es el primero en salir | FIFO — el primero en entrar es el primero en salir | FIFO, pero con **cupo fijo** |
| **Por dónde entra** | por la cima | por el final | por el final |
| **Por dónde sale** | por la cima | por el frente | por el frente |
| **¿Crece?** | sí | sí | **no**: la capacidad se fija al crearla |
| **Cuando se llena** | no se llena | no se llena | rechaza (o pisa el más viejo) |
| **Se usa para** | deshacer, recursión, expresiones, *backtracking* | turnos, procesos, BFS | buffers, últimas N lecturas, Round-Robin |

# 2. Los métodos, lado a lado

## 2.1 Pila

| Método | Qué hace | Si está vacía |
|---|---|---|
| `apilar(elemento)` | pone un elemento en la cima | — |
| `desapilar()` | saca y retorna el de la cima | lanza `IndexError` |
| `cima` *(propiedad)* | retorna el de la cima **sin sacarlo** | lanza `IndexError` |
| `esta_vacia()` | `True` si no hay elementos | — |
| `len(p)` | cuántos elementos hay | `0` |
| `bool(p)` / `if p:` | `True` si tiene elementos | `False` |
| `for x in p` | recorre **de la cima hacia la base** | no itera |
| `x in p` | `True` si el elemento está | `False` |
| `repr(p)` | texto legible para depurar | — |

## 2.2 Cola

| Método | Qué hace | Si está vacía |
|---|---|---|
| `encolar(elemento)` | pone un elemento al final | — |
| `desencolar()` | saca y retorna el del frente | lanza `IndexError` |
| `frente` *(propiedad)* | retorna el del frente **sin sacarlo** | lanza `IndexError` |
| `final` *(propiedad)* | retorna el del final **sin sacarlo** | lanza `IndexError` |
| `esta_vacia()` | `True` si no hay elementos | — |
| `len(c)` | cuántos elementos hay | `0` |
| `bool(c)` / `if c:` | `True` si tiene elementos | `False` |
| `for x in c` | recorre **del frente hacia el final** | no itera |
| `x in c` | `True` si el elemento está | `False` |
| `repr(c)` | texto legible para depurar | — |

## 2.3 Cola circular

Todo lo de la cola, **más** el manejo del cupo:

| Método | Qué hace | Casos límite |
|---|---|---|
| `ColaCircular(capacidad)` | crea la cola con cupo fijo | `ValueError` si la capacidad es menor que 1 |
| `capacidad` *(propiedad)* | cuántos caben en total | **solo lectura**: no se puede reasignar |
| `encolar(elemento)` | pone un elemento al final | lanza `OverflowError` si está llena |
| `desencolar()` | saca y retorna el del frente | lanza `IndexError` si está vacía |
| `frente` *(propiedad)* | el del frente, sin sacarlo | lanza `IndexError` si está vacía |
| `esta_llena()` | `True` si ya no cabe ni uno más | — |
| `esta_vacia()` | `True` si no hay elementos | — |
| `len(c)` | cuántos hay **ahora** (no la capacidad) | `0` |
| `for x in c` | recorre en orden **lógico**, no físico | — |

## 2.4 Lo que cuesta cada operación

| Operación | Pila (`list`) | Cola (`list`) | Cola (`deque`) | Cola circular |
|---|---|---|---|---|
| Meter | O(1) | O(1) | O(1) | O(1) |
| Sacar | O(1) | **O(n)** ← `pop(0)` | O(1) | O(1) |
| Consultar extremo | O(1) | O(1) | O(1) | O(1) |
| Memoria | crece | crece | crece | **fija** |

```{note}
La celda `O(n)` de la segunda columna es la razón por la que tu clase `Cola` **no** debe usar
`lista.pop(0)`. Es el mismo hallazgo del ejercicio de la ruta del bus.
```

## Banco de pruebas

Ejecuta esta celda **una vez** antes de empezar. No la modifiques.

````python
# ============ BANCO DE PRUEBAS (identico al del cuaderno) ============
def _probar(nombre, casos):
    ok = 0
    for etiqueta, fn in casos:
        try:
            fn(); print(f"  OK     {etiqueta}"); ok += 1
        except AssertionError as e:
            print(f"  FALLA  {etiqueta}  -> {e or 'el resultado no es el esperado'}")
        except Exception as e:
            print(f"  ERROR  {etiqueta}  -> {type(e).__name__}: {e}")
    print(f"\n  {nombre}: {ok}/{len(casos)} pruebas superadas\n")

def _lanza(excepcion, fn):
    try: fn()
    except excepcion: return
    except Exception as e: raise AssertionError(f"lanzo {type(e).__name__} en vez de {excepcion.__name__}")
    raise AssertionError(f"no lanzo {excepcion.__name__}")

def probar_pila(Pila):
    def construccion():
        assert len(Pila()) == 0
        assert len(Pila([1, 2, 3])) == 3
        assert Pila([1, 2, 3]).cima == 3, "el ultimo del iterable debe quedar en la cima"
    def lifo():
        p = Pila(); p.apilar("A"); p.apilar("B"); p.apilar("C")
        assert p.desapilar() == "C" and p.desapilar() == "B" and p.desapilar() == "A"
    def cima_no_saca():
        p = Pila([1, 2]); assert p.cima == 2 and len(p) == 2, "cima no debe eliminar"
    def vacias():
        p = Pila()
        assert p.esta_vacia() and not p and len(p) == 0
        _lanza(IndexError, p.desapilar)
        _lanza(IndexError, lambda: p.cima)
    def protocolos():
        p = Pila([1, 2, 3])
        assert bool(p) is True and 2 in p and 9 not in p
        assert list(p) == [3, 2, 1], "__iter__ debe ir de la cima hacia la base"
        assert "Pila" in repr(p), "__repr__ debe nombrar la clase"
    def encapsulado():
        p = Pila([1, 2])
        assert not hasattr(p, "datos"), "la lista interna debe llamarse _datos, no datos"
    _probar("Pila", [("construccion", construccion), ("orden LIFO", lifo),
                     ("cima no elimina", cima_no_saca), ("pila vacia", vacias),
                     ("protocolos", protocolos), ("encapsulamiento", encapsulado)])

def probar_cola(Cola):
    def construccion():
        assert len(Cola()) == 0 and len(Cola([1, 2, 3])) == 3
        assert Cola([1, 2, 3]).frente == 1, "el primero del iterable debe quedar al frente"
    def fifo():
        c = Cola(); c.encolar("A"); c.encolar("B"); c.encolar("C")
        assert c.desencolar() == "A" and c.desencolar() == "B" and c.desencolar() == "C"
    def extremos():
        c = Cola([1, 2, 3])
        assert c.frente == 1 and c.final == 3 and len(c) == 3, "frente/final no deben eliminar"
    def vacias():
        c = Cola()
        assert c.esta_vacia() and not c
        _lanza(IndexError, c.desencolar)
        _lanza(IndexError, lambda: c.frente)
        _lanza(IndexError, lambda: c.final)
    def protocolos():
        c = Cola([1, 2, 3])
        assert 2 in c and 9 not in c
        assert list(c) == [1, 2, 3], "__iter__ debe ir del frente hacia el final"
        assert "Cola" in repr(c)
    _probar("Cola", [("construccion", construccion), ("orden FIFO", fifo),
                     ("frente y final", extremos), ("cola vacia", vacias),
                     ("protocolos", protocolos)])

def probar_cola_circular(ColaCircular):
    def construccion():
        c = ColaCircular(3)
        assert c.capacidad == 3 and len(c) == 0 and c.esta_vacia()
        _lanza(ValueError, lambda: ColaCircular(0))
    def llenado():
        c = ColaCircular(3)
        for x in "ABC": c.encolar(x)
        assert c.esta_llena() and len(c) == 3
        _lanza(OverflowError, lambda: c.encolar("D"))
    def da_la_vuelta():
        c = ColaCircular(4)
        for x in "ABCD": c.encolar(x)
        assert c.desencolar() == "A" and c.desencolar() == "B"
        c.encolar("E"); c.encolar("F")
        assert list(c) == ["C", "D", "E", "F"], f"orden logico incorrecto: {list(c)}"
        assert len(c) == 4 and c.esta_llena()
    def capacidad_fija():
        c = ColaCircular(2)
        for _ in range(20):
            c.encolar(1); c.desencolar()
        assert c.capacidad == 2, "la capacidad no puede cambiar"
        assert len(c._datos) == 2, "el arreglo interno no debe crecer"
    def vacias():
        c = ColaCircular(2)
        _lanza(IndexError, c.desencolar)
        _lanza(IndexError, lambda: c.frente)
    def solo_lectura():
        c = ColaCircular(3)
        _lanza(AttributeError, lambda: setattr(c, "capacidad", 99))
    _probar("ColaCircular", [("construccion", construccion), ("se llena y rechaza", llenado),
                             ("da la vuelta", da_la_vuelta), ("capacidad fija", capacidad_fija),
                             ("cola vacia", vacias), ("capacidad de solo lectura", solo_lectura)])
````
